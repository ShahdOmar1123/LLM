```python
prompt="Suggest 3 innovative names for a student study mobile app, formatted as a numbered list, and explain each in one line"
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
)
```

```python
Profile1=client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=0.0,

    )
)
```
